```python
from langchain_community.document_loaders import PyPDFLoader


loader = PyPDFLoader("./UNIT 2 E Commerce.pdf")

docs = loader.load()

docs
```

```python
## STEP 2: SPLITTING THE DOCUMENTS INTO CHUNKS

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)
splits = text_splitter.split_documents(docs)

print(f"Total chunks created: {len(splits)}")
splits
```

```python
len(splits)
```

```python
## STEP 4: CREATE AND STORE EMBEDDINGS IN VECTRE STORE

from langchain_community.document_loaders import PyPDFLoader

# Load the solar energy PDF file from your project root directory
loader = PyPDFLoader("UNIT 2 E Commerce.pdf")
pages = loader.load()

print(f"Total pages loaded: {len(pages)}")
print(pages[0].page_content[:300])
```

```python
## STEP 5: SEMANTIC SEARCH

from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma

# Use the active multimodal preview model identifier supported by the SDK
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")

vectorstore = Chroma.from_documents(
    documents=splits,
    embedding=embeddings
)

print("vectorstore is successfully defined!")
```

```python
## TALK TO LLM

query = "What are the main topics in e-commerce?"
results = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(results, 1):
    print(f"Result {i}:\n{doc.page_content}\n" + "-" * 40)
```
